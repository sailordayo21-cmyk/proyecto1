```python
#Scararemos los nombre de las columnas numericas y categoricas.
def tipo_columnas_ordenadas(df):  # Creo una función
    cat = []
    num = []

    for col in df.columns:
        if df[col].dtype == 'object':
            cat.append(col)
        else:
            num.append(col)

    return cat, num

# --- Forma correcta de llamarla e imprimir los resultados ---
cat, num = tipo_columnas_ordenadas(df)
print('Las Columnas Categoricas son:', cat)
print('Las Columnas Numéricas son:', num)


```

```python
#Respondermos a la pregunta de Cuantas personas han sobrevivido?
df['Survived'].value_counts()
```

Reflexión: Podemos decir que menos de la midad sobrevivieron al accidente.

```python
import matplotlib.pyplot as plt
supervivientes = df['Survived'].value_counts()
supervivientes_columnas = ['NO Sobrevivió', 'SI Sobrevivió']

#Empezamos a graficar
fig, ax = plt.subplots() # Create figure and axes without figsize argument
fig.set_size_inches(5, 5) # Set the figure size explicitly after creation
ax.bar(supervivientes_columnas, supervivientes, color=['red', 'blue']) # Plot on the axes
ax.set_title('Número de personas que sobrevivieron en el Titanic', fontsize=15)
ax.set_xlabel('Supervivencia')  # etiqueta en el eje de las x
ax.set_ylabel('Número de personas') # etiqueta en el eje de las x
plt.show()
```

```python
#creamos una variables para conocer los el porcentaje
pasajeros_totales=len(df)
mujeres = df[df['Sex']== 'female']
hombres = df[df['Sex']=='male']
porcentaje_mujeres = (len(mujeres)/pasajeros_totales)*100
porcentaje_hombres = (len(hombres)/pasajeros_totales)*100

#print('hola', porcentaje_mujeres)

print(f'Mujeres:{porcentaje_mujeres:.2f}%')
print(f'Hombre:{porcentaje_hombres:.2f}%')

```

```python
#Graficaremos el porcentaje
porcentaje_genero = pd.DataFrame({'Genero':['Mujeres','Hombres'],'Porcentaje':[porcentaje_mujeres,porcentaje_hombres]})
#porcentaje_genero = porcentaje_genero.set_index('Genero')
#porcentaje_genero

#graficamos
fig = px.pie(porcentaje_genero, names='Genero', values='Porcentaje' ,title='Porcentaje de Mujeres y Hombres')
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(showlegend=False)
fig.show()
```

En el barco Titanic predomina más el genero masculino, con un porcentahe de (64.8%), frente a los (35.2%) del genero femenino, esta  seria la posible razón principal de la epoca, ya que más de 100 años que se embarcaron

```python
#primero agruparemos la columnas sexo y supervivientes
Superviviente_por_genero = df.groupby(['Sex','Survived']).size().unstack()
#Ahora calculamos el porcentaje de supervivientes por genero
Porcentaje_supervivientes = (Superviviente_por_genero[1]/(Superviviente_por_genero[0]+Superviviente_por_genero[1]))*100
sns.barplot(x=Porcentaje_supervivientes.index, y=Porcentaje_supervivientes.values, palette={'male':'blue', 'female':'pink'})
plt.xlabel('Género')
plt.ylabel('Porcentaje de supervivinetes')
plt.title('Porcentaje de supervivientes por género', fontsize=16)
plt.show()
```

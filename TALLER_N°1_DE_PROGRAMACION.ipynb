1) Mete los valores del 1 al 100 en una lista, e imprimirla y así miso identificar cual es máximo , minino y el promedio.



```python
#1) Mete los valores del 1 al 100 en una lista, e imprimirla y así miso identificar cual es máximo , minino y el promedio.
numeros = list(range (1,100, 1))
print(numeros)

maximo=max(numeros)
minimo=min(numeros)
promedio = sum(numeros) / len(numeros)

print("El valor Máximo:", maximo)
print("El valor Mínimo:", minimo)
print("Promedio de los numeros del 1 al 100 es :", promedio)
```

6) Crea un diccionario donde la clave sea el nombre del usuario y el valor sea el teléfono (no es necesario validar). Tendrás que ir pidiendo contactos hasta el usuario diga que no quiere insertar mas. No se podrán meter nombres repetidos.

```python
# Diccionario de contactos , sin nombres repetidos
contactos = {}
while True:
    respuesta = input("quieres agregar un contacto? (si/no):")
    if respuesta.lower() != "si":
        break
    nombre = input("ingrese el nombre del usuario:")
    if nombre in contactos:
        print("Ese nombre ya existe, no se puede repetir")
    else:
        telefono = input("Ingrese el telefono:")
        contactos[nombre] = telefono
        print(f"Contacto {nombre} agregado correctamente")
print("Lista final de contactos:")
print(contactos)
```

9) Pide una cadena por teclado, mete los caracteres en una lista sin repetir caracteres.

```python
# Pedir una cadena, guardar sus caracteres en una lista sin repetir

cadena = input("Ingrese una cadena de texto: ")

caracteres = []

for letra in cadena:
    # Solo agregamos la letra si todavía no está en la lista
    if letra not in caracteres:
        caracteres.append(letra)

print("Caracteres sin repetir:")
print(caracteres)
```
